```python
import numpy as np
```

```python
a = np.random.randint(1 , 20, (3,4))
b = np.random.randint(1 , 20, (3,4))
print("matrix a:")
print(a)
print("matrix b:")
print(b)
```

```python
mean_a = np.mean(a, axis=1)
mean_b = np.mean(b, axis=1)
print("the mean of each row in a is:")
print(mean_a)
print("the mean of each row in b is:")
print(mean_b)
```

```python
new_a = a + mean_b[:, np.newaxis]
new_b = b + mean_a[:, np.newaxis]

print("new matrix a:")
print(a)

print("new matrix b:")
print(b)
```
