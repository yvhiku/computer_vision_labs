# 1. NumPy

NumPy is a library that provides implementations of algorithms optimized for working with multidimensional arrays. NumPy is based on the LAPACK library, which makes computations fast.

In addition to its core functionality (multidimensional arrays), NumPy includes a set of packages for specialized tasks, for example:

- numpy.linalg — implements linear algebra operations (basic vector and matrix multiplication is available in the core functionality);
- numpy.random — provides functions for working with random variables.

NumPy Array vs. Python List:
- NumPy is faster than ordinary lists when operations can be vectorized.
- Adding elements to an existing array is slower in NumPy than in a Python list. Adding an element to a NumPy array has O(N) complexity, whereas in a Python list it is O(1).
- NumPy arrays can contain elements of only one data type.

The NumPy documentation is available [here](https://numpy.org/devdocs/reference/index.html#reference)

### 1.2.2 Creating Empty Arrays and Arrays Filled with Values

NumPy arrays cannot be extended in the same way as Python lists because no extra space is reserved at the end of an array for quick additions. Therefore, common practice is either to extend a Python list and convert it to a NumPy array, or to allocate the required space in advance using `np.zeros` or `np.empty`.

```python
a = np.zeros(3, int)  # one-dimensional array of zeros, length 3
b = np.empty(3, int)  # empty one-dimensional array, length 3
c = np.ones(3)  # array filled with ones
d = np.full(3, 7.)  # array filled with the value 7

e = np.zeros_like(a)  # create a zero array with the same shape and data type as 'a'
f = np.empty_like(b)  # create an empty array with the same shape and data type as 'b'
g = np.ones_like(c)  # create an array of ones with the same shape and data type as 'c'
h = np.full_like(d, 8)  # create a filled array with the same shape and data type as 'd'
print(b)
```

To create an array, specify its two dimensions using a tuple (in parentheses):

```python
k = np.ones((3, 2))
print(k)
```

Similarly, the `np.zeros()`, `np.empty()`, and `np.full()` methods, as well as their corresponding functions `np.zeros_like()`, `np.empty_like()`, `np.ones_like()`, and `np.full_like()`, can be used with matrices.

The `np.eye()` method creates an identity matrix:

```python
a = np.eye(3, 3)
print(a)
```

**Initializing an array with a sequence of numbers** can be done using the `np.arange()` and `np.linspace()` methods.

`np.arange(stop)` — creates an array containing a sequence of numbers from 0 up to `stop`.
`np.arange(start, stop)` — creates an array containing numbers from `start` up to `stop`.
`np.arange(start, stop, step)` — creates an array containing numbers from `start` up to `stop`, with the specified `step`.

`np.linspace(start, stop, n)` — creates an array of length `n` with values ranging from `start` to `stop`. The step size is calculated automatically.

```python
a = np.arange(6)
b = np.arange(2, 6)
c = np.linspace(0, 0.5, 6)

print('a =', a)
print('b =', b)
print('c =', c)
```

It is also possible to create an array filled with random values:

`rnd = np.random.default_rng()` — creates a random number generator.
`rnd.integers(min, max, n)` — returns an array of length `n` containing uniformly distributed random integers, where x ∈ [min, max).
`rnd.integers(min, max, n, endpoint=True)` — returns an array of length `n` containing uniformly distributed random integers, where x ∈ [min, max].
`rnd.random(n)` — returns an array of length `n` containing uniformly distributed random floating-point numbers in the interval [0, 1).
`rnd.uniform(min, max, n)` — returns an array of length `n` containing uniformly distributed random floating-point numbers in the interval [min, max).
`rnd.standard_normal(n)` — returns an array of length `n` containing normally distributed random floating-point numbers, where μ=0 and σ=1.
`rnd.normal(μ, σ, n)` — returns an array of length `n` containing normally distributed random floating-point numbers with the specified μ and σ.

```python
rnd = np.random.default_rng()
a = rnd.integers(0, 10, 3)
b = rnd.integers(0, 10, 3, endpoint=True)
c = rnd.random(3)
d = rnd.uniform(1, 10, 3)
e = rnd.standard_normal(3)
f = rnd.normal(5, 2, 3)
```

In the same way, a matrix filled with random numbers can be created. Its dimensions are also specified using a tuple (in parentheses):

```python
rnd = np.random.default_rng()
a = rnd.integers(0, 10, (3, 2))
b = rnd.integers(0, 10, (3, 2), endpoint=True)
c = rnd.random((3, 2))
d = rnd.uniform(1, 10, (3, 2))
e = rnd.standard_normal((3, 2))
f = rnd.normal(5, 2, (3, 2))
```

### 1.3.1 Copying Array Elements
Remember that a NumPy array is an object. Therefore, simply assigning an array to a new variable does not copy the array's values; it copies a reference to the array object.
Consider the following example:

```python
a = np.array([1, 2, 3])
b = a  # this copies a reference, not the array itself!
b[0] = 10  # Now change the first array element to 10 and print the contents of a and b.
print('a =', a)
print('b =', b)
```

To copy an array, use the **copy()** method.

```python
a = np.array([1, 2, 3])
b = a.copy()
b[0] = 10
print('a =', a)
print('b =', b)
```

## 1.5 Matrix Operations
NumPy also supports matrix operations.
The following operations are worth noting:
- `A * B` — element-wise multiplication;
- `A @ B` — matrix multiplication;
- `A / B` — element-wise division.

```python
a = np.array([[1, 2], [3, 4]])
b = np.array([[2, 0], [0, 2]])

print('b + a =', b + a, sep='\n')
print('b - a =', b - a, sep='\n')
print('b * a =', b * a, sep='\n')
print('b @ a =', b @ a, sep='\n')
print('b / a =', b / a, sep='\n')
print('b // a =', b // a, sep='\n')
```

### 1.5.1 Transposition
A matrix can be transposed in two ways:
`a.transpose()` or `a.T`

**Transposing vectors works differently.** If you try to transpose a row vector, for example:
    `a = np.array([1, 2, 3])`
    `a.T`
nothing will change. The vector remains a one-dimensional array. This happens because array `a` has only one dimension.

The simplest way to transpose a vector is to change its shape:
`a.reshape(-1, 1)`
Here, `-1` tells NumPy to calculate the dimension automatically.

```python
a = np.array([1, 2, 3])
b = a.reshape(-1, 1)
print(b)
print('a * b =\n', a * b)
```

#### 1.5.3.2 Array Splitting Functions
The reverse operation of array (vector) concatenation is array splitting, which can be performed using the `hsplit()` and `vsplit()` methods.

```python
a = np.arange(1, 13).reshape(3, -1)
b = np.arange(1, 9).reshape(2, -1)
c = np.arange(1, 7).reshape(3, -1)
d = np.vstack((a, b))
e = np.hstack((a, c))
print('vsplit(d,[3])) =', np.vsplit(d, [3]), sep='\n')
print('hsplit((e,[4]])) =', np.hsplit(e, [4]), sep='\n')
```

## 1.7 Data Type Conversion in NumPy
When working with images, video, or simply large matrices, precise control over data types may be required. Unlike standard Python, which provides only a few numeric types, NumPy offers a wider range of data types.

The following data types are available in NumPy:

- `bool_` — Boolean type (True or False), stored as a byte.
- `int_` — default integer type (the same as C `long`; usually either int64 or int32).
- `intc` — identical to C `int` (int32 or int64).
- `intp` — integer type used for indexing (the same as C `ssize_t`; usually either int64 or int32).
- `int8` — integers in the range from -128 to 127 (1 byte).
- `int16` — integers in the range from -32768 to 32767 (2 bytes).
- `int32` — integers in the range from -2147483648 to 2147483647 (4 bytes).
- `int64` — integers in the range from -9223372036854775808 to 9223372036854775807 (8 bytes).
- `uint8` — integers in the range from 0 to 255 (1 byte).
- `uint16` — integers in the range from 0 to 65535 (2 bytes).
- `uint32` — integers in the range from 0 to 4294967295 (4 bytes).
- `uint64` — integers in the range from 0 to 18446744073709551615 (8 bytes).
- `float_` — the same as float64.
- `float16` — half-precision floating-point numbers: 1 sign bit, 5 exponent bits, and 10 mantissa bits (2 bytes).
- `float32` — single-precision floating-point numbers: 1 sign bit, 8 exponent bits, and 23 mantissa bits (4 bytes).
- `float64` — double-precision floating-point numbers: 1 sign bit, 11 exponent bits, and 52 mantissa bits (8 bytes).
- `complex_` — the same as complex128.
- `complex64` — complex numbers whose real and imaginary parts are represented by two float32 values.
- `complex128` — complex numbers whose real and imaginary parts are represented by two float64 values.

The complete list of types can be viewed by displaying the contents of the `np.sctypeDict` dictionary.

```python
np.sctypeDict
```

If NumPy is imported as `import numpy as np`, data type objects become available as `np.bool_`, `np.float64`, `np.int16`, etc.

To convert one data type to another, you can use data type constructors:
- `np.float64(11)`
- `np.int64(11)`
- `np.float64([1, 2, 3])`
- `np.int16([1, 2, 3])`

Alternatively, you can use the `astype()` method:

```python
a = np.float64([1, 2, 3])
a.astype(np.uint8)
```

If you need information about a specific floating-point type, you can obtain it using the `finfo()` function:
- `np.finfo(np.float64)`
- `np.finfo(np.float)`
- `np.finfo(np.float16)`

```python
np.finfo(np.float64)
```
